# Extracting model info and Information Criterion (IC) from hddm

Below is a iPython notebook for extracting details about all models and information criteria for each model. 

Some useful notes:
- Make sure to specify paths to your model files (path_to_json) and output files from hddm (path_to_ICs). 
- Double check the outputs, as you may need to adjust some of the extracted data (e.g., model names when extracting ICs based on your specific filenames). 
- When extracting model information, you might want more/less informationextracted (e.g., bias = info['bias]).

## Import packages and specify paths

```python
import os, glob, re, json
import numpy as np
import pandas as pd 

# Path to the models
path_to_json = '/gpfs/data/ashenhav/mri-data/TCB/manuscript/analysis/ddm/models/'

# Path to the HDDM outputs
path_to_ICs = '/gpfs/data/ashenhav/mri-data/TCB/manuscript/analysis/ddm/batch/logs/'

# Path to where to save the output
path_to_save = '/gpfs/data/ashenhav/mri-data/TCB/manuscript/analysis/ddm/output/errors_automatic/rewardpenalty/'
```

```python
print(path_to_json)
```

## Extract the model information

```python
# get list of .json files 
os.chdir(path_to_json)
files = glob.glob('*.json')
json_files = [pos_json for pos_json in os.listdir(path_to_json) if pos_json.endswith('.json')]

# create my dataframe of modelinfo
modelSummary = pd.DataFrame(columns=['Study', 'ModelName','dataFile','errorInfo', 'formula', 'a','v','z', 'bias','intertrialvariability','totalSample','burnedsample'])

# loop through json files  extract info
for i in range(len(files)): 
    
    info = open(files[i])
    info = json.load(info)
    modelName = info['model']
    study = info['study']
    datafile = info['data']
    errorInfo = info['errors']
    formula = info['formula']
    formula2=[formula]
    totalSample= info['totalsample']
    burnedsample=info['burnedsample']
    intertrialvariability=info['includes']
    bias = info['bias']
    curList=[study,modelName,datafile,errorInfo,formula2]

    for n in range(len(formula)):
        curFormula=formula[n]
        currentParameter =curFormula.split("~",1)
        parameter=currentParameter[0]
        parameterFormula=currentParameter[1]
        curList.append(parameterFormula)

    if len(formula)==2:
        curList.append("NA")

    curList.append(bias)
    curList.append(intertrialvariability)
    curList.append(totalSample)
    curList.append(burnedsample)
    modelSummary.loc[i]=curList

modelSummary=modelSummary.sort_values(by=['ModelName'])
modelSummary=modelSummary.reset_index()
modelSummary=modelSummary.drop(['index'], axis=1)

print(modelSummary.iloc[:, : 4])
```

## Extract the information criteria

```python
# Get the list of .out files which cointain the DIC/BPIC info
os.chdir(path_to_ICs)
files = glob.glob('ppc*')

# Initialize the results matrix
results_IC = np.empty([len(files), 3], dtype="S30")
modelIC = pd.DataFrame(columns=['ModelName','DIC','BPIC'])

# Loop over the files
for i in range(len(files)):
    
    # Read in the log file into model object
    model = open(files[i],'r')
    model = model.read()
    
    Model_index = model.find("ModelName:") 
    DIC_index = model.find("DIC:")
    BPIC_index = model.find("BPIC:")
    
    ModelName = model[Model_index:(Model_index+18)]
    
    if ModelName[-1] is ('_'):
        ModelName = ModelName[-7:-1]
    else:
        ModelName = ModelName[-7:]
        
    print(ModelName)
    
    
    DIC = model[DIC_index:(DIC_index+18)]
    DIC = re.sub(r'^DIC: ','',DIC)
    
    BPIC = model[BPIC_index:(BPIC_index+18)]
    BPIC = re.sub(r'^BPIC: ','',BPIC)
    
    curList=[ModelName,DIC,BPIC]
    #print(curList)
    
    modelIC.loc[i]=curList

# Sort by Model Names and reset index
modelIC=modelIC.sort_values(by=['ModelName'])
modelIC=modelIC.reset_index()
modelIC=modelIC.drop(['index'], axis=1)

# Visualize the Model IC results
modelIC
```

```python
# Visualize Model IC results, sorted by BPIC
modelIC.sort_values(by='BPIC')
```

```python
# Visualize Model IC results, sorted by BPIC
modelIC.sort_values(by='DIC')
```

## Merge the datasets and save as a csv

Merged the two data frames by the 'ModelNum' variable

```python
# split the ModelName variable by model number
modelSummary[['ModelNum','ModelType','SRT']] = modelSummary.ModelName.str.split("_", expand = True)

# rename ModelName to ModelNum
modelIC = modelIC.rename(columns={'ModelName': 'ModelNum'})

# join datasets by ModelNum
results=modelSummary.join(modelIC.set_index('ModelNum'), on='ModelNum')

# save to csv
results.to_csv((path_to_save+"InformationCriteria.csv"), index=False)

# visualize
results.sort_values(by="BPIC")
```
